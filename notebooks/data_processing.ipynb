```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("Dingdong-Inc/FreshRetailNet-50K")
df = ds["train"].to_pandas()
print(df.shape)
df.head()
```

```python
#shrinking data into a sample:

stores = df["store_id"].unique()[:20]
df = df[df["store_id"].isin(stores)].copy()
df["dt"] = pd.to_datetime(df["dt"])
df = df.drop_duplicates(subset=["store_id", "product_id", "dt"])
df = df.sort_values(["store_id", "product_id", "dt"]).reset_index(drop=True)
print(df.shape)
```

```python
# Checks
list_cnt = df["hours_stock_status"].apply(lambda x: sum(x[6:22]))
print("count mismatches:", (list_cnt != df["stock_hour6_22_cnt"]).sum())

hourly_total = df["hours_sale"].apply(sum)
print("sales mismatches:", ((hourly_total - df["sale_amount"]).abs() > 0.05).sum())

s = df.groupby(["store_id", "product_id"])["dt"].agg(["min", "max", "count"])
s["expected"] = (s["max"] - s["min"]).dt.days + 1
print("series with gaps:", (s["count"] != s["expected"]).sum(), "of", len(s))

print(df["sale_amount"].describe(percentiles=[.5, .95, .99]))
print(df[["discount", "precpt", "avg_temperature", "avg_humidity", "avg_wind_level"]].describe())
```

```python
df["stockout_hours"] = df["stock_hour6_22_cnt"]
g_keys = [df["store_id"], df["product_id"]]

for horizon in [1, 3, 7]:
    for min_hours in [1, 4, 8]:
        s = (df["stockout_hours"] >= min_hours).astype(int)
        g = s.groupby(g_keys)
        future = pd.concat([g.shift(-i) for i in range(1, horizon + 1)], axis=1)
        valid = future.notna().all(axis=1)
        rate = future[valid].max(axis=1).mean()
        print(f"horizon={horizon}d, min_hours={min_hours}: {rate:.1%} positive")
```

```python
HORIZON, MIN_HOURS = 3, 8
keys = ["store_id", "product_id"]

# Today's serious-stockout flag
df["stockout_today"] = (df["stock_hour6_22_cnt"] >= MIN_HOURS).astype(int)

# Target: serious stockout on any of the next 3 days
g = df.groupby(keys)["stockout_today"]
future = pd.concat([g.shift(-i) for i in range(1, HORIZON + 1)], axis=1)
df["target"] = future.max(axis=1)
df = df[future.notna().all(axis=1)].copy()
df["target"] = df["target"].astype(int)

# Features (only information known today or earlier)
df = df.drop(columns=["stockout_hours"], errors="ignore")
gs = df.groupby(keys)["sale_amount"]
df["sales_lag1"] = gs.shift(1)
df["sales_avg7"] = gs.transform(lambda s: s.rolling(7).mean())
df["stockout_rate7"] = df.groupby(keys)["stockout_today"].transform(lambda s: s.rolling(7).mean())
df["day_of_week"] = df["dt"].dt.dayofweek
df["month"] = df["dt"].dt.month
df = df.dropna(subset=["sales_lag1", "sales_avg7", "stockout_rate7"]).reset_index(drop=True)

print(df.shape)
print(df["target"].value_counts(normalize=True))
```

```python
#data visualisation:
import matplotlib.pyplot as plt
import seaborn as sns

fig, ax = plt.subplots(2, 2, figsize=(13, 10))
df["target"].value_counts().plot.bar(ax=ax[0, 0], title="Class balance (target)")
df["sale_amount"].clip(upper=df["sale_amount"].quantile(.99)).plot.hist(
    bins=40, ax=ax[0, 1], title="Daily sales (clipped at 99th percentile)")
df.groupby("day_of_week")["target"].mean().plot.bar(
    ax=ax[1, 0], title="Stockout risk by weekday (0 = Monday)")
num = ["sale_amount", "stock_hour6_22_cnt", "discount", "precpt", "avg_temperature",
       "avg_humidity", "avg_wind_level", "sales_lag1", "sales_avg7", "stockout_rate7", "target"]
sns.heatmap(df[num].corr(), cmap="coolwarm", center=0, annot=True, fmt=".2f",
            annot_kws={"size": 6}, ax=ax[1, 1])
ax[1, 1].set_title("Correlations")
plt.tight_layout()
plt.savefig("../report/eda_overview.png", dpi=150)
plt.show()
```

```python
#preprocessing, time-based split and scaling data:

from sklearn.preprocessing import StandardScaler
import joblib

# Drop list columns, redundant flag and high-cardinality ID columns
drop_cols = ["hours_sale", "hours_stock_status", "stockout_today", "city_id", "store_id",
             "management_group_id", "second_category_id", "third_category_id", "product_id"]
model_df = df.drop(columns=drop_cols)

# Encode categorical features
model_df = pd.get_dummies(model_df, columns=["first_category_id", "day_of_week"], dtype=int)

# Time-based split: train on the past, test on the future (3-day gap = target window)
cut = model_df["dt"].quantile(0.8)
train = model_df[model_df["dt"] <= cut - pd.Timedelta(days=3)].copy()
test = model_df[model_df["dt"] > cut].copy()

# Scale numeric features: fit on TRAIN only to avoid leakage
num_cols = ["sale_amount", "stock_hour6_22_cnt", "discount", "precpt", "avg_temperature",
            "avg_humidity", "avg_wind_level", "sales_lag1", "sales_avg7", "stockout_rate7", "month"]
scaler = StandardScaler().fit(train[num_cols])
train[num_cols] = scaler.transform(train[num_cols])
test[num_cols] = scaler.transform(test[num_cols])

train.to_csv("../data/train_processed.csv", index=False)
test.to_csv("../data/test_processed.csv", index=False)
joblib.dump(scaler, "../data/scaler.joblib")

print(train.shape, test.shape)
print(train["target"].mean(), test["target"].mean())
```
