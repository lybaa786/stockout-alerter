```python
from datasets import load_dataset
import pandas as pd

ds = load_dataset("Dingdong-Inc/FreshRetailNet-50K")
df = ds["train"].to_pandas()
print(df.shape)
df.head()
```

```python
#shrinking data into a sample:

stores = df["store_id"].unique()[:20]
df = df[df["store_id"].isin(stores)].copy()
df["dt"] = pd.to_datetime(df["dt"])
df = df.drop_duplicates(subset=["store_id", "product_id", "dt"])
df = df.sort_values(["store_id", "product_id", "dt"]).reset_index(drop=True)
print(df.shape)
```

```python
#building target for the model:

df["stockout_day"] = (df["stock_hour6_22_cnt"] > 0).astype(int)
g = df.groupby(["store_id", "product_id"])["stockout_day"]
future = pd.concat([g.shift(-i) for i in range(1, 8)], axis=1)
df["stockout_next7"] = future.max(axis=1)
df = df[future.notna().all(axis=1)].copy()   # drop last 7 days of each series
df["stockout_next7"] = df["stockout_next7"].astype(int)
df["stockout_next7"].value_counts(normalize=True)
```
